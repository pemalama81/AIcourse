```python
#Importing libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Load Dataset
df = sns.load_dataset("titanic")
```

```python
df_copy = df.copy()
```

```python
df
```

```python
df["age"].isnull().sum()
```

```python
sns.histplot(df["age"])
```

```python
#Box plot to check outliers on Column "Age"
sns.boxplot(df["age"])
```

```python
df.isnull().sum()
```

```python
#Data cleaning
df_copy["age"].fillna(df_copy["age"].median(),inplace = True)
```

```python
df_copy["age"].isnull().sum()
```

```python
df_copy.isnull().sum()
```

```python
df_copy.drop(columns=['deck'],inplace = True)
```

```python
df_copy
```

```python
df_copy["embarked"].unique()
```

```python
df_copy["embarked"].fillna(df_copy["embarked"].mode()[0],inplace = True)
```

```python
df_copy.isnull().sum()
```

```python
#Encoding Categorical values
df_copy["sex"] = df_copy["sex"].map({"male": 0, "female":1})
```
